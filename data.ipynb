```python
import pandas as pd
```

```python

# 1. 엑셀 파일(.xlsx) 불러오기
#    파일명이 'data.xlsx', 시트명이 기본(첫 번째)일 경우
df = pd.read_excel('전주공공와이파이.xlsx', engine='openpyxl')

# 2. latitude와 longitude 두 컬럼이 모두 동일한 행(중복 장소) 제거
df_unique = df.drop_duplicates(subset=['latitude', 'longitude'], keep='first')

# 3. 결과를 CSV 파일로 저장
#    인덱스는 필요 없으면 index=False
# CP949(혹은 ANSI) 인코딩으로 저장
df_unique.to_csv('wifi.csv', index=False, encoding='cp949')



# 확인
print(f"원본 행 개수: {len(df)}")
print(f"중복 제거 후 행 개수: {len(df_unique)}")
```
